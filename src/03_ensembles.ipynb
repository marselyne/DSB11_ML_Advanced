## 0. Imports

````python
import pandas as pd
from sklearn.model_selection import train_test_split, cross_validate, StratifiedKFold, GridSearchCV
from sklearn.metrics import accuracy_score, precision_score, recall_score
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, VotingClassifier, BaggingClassifier, StackingClassifier
from sklearn.multiclass import OneVsRestClassifier
import joblib
````

````python
import warnings
warnings.filterwarnings("ignore")
````

## 1. Preprocessing

1. Create the same dataframe as in the previous exercise.
2. Using `train_test_split` with parameters `test_size=0.2`, `random_state=21` get `X_train`, `y_train`, `X_test`, `y_test` and then get `X_train`, `y_train`, `X_valid`, `y_valid` from the previous `X_train`, `y_train`. Use the additional parameter `stratify`.

````python
df = pd.read_csv('data/day-of-week-not-scaled.csv')
df.head(5)
````

````python
df_1 = pd.read_csv('data/dayofweek.csv')
df['dayofweek'] = df_1['dayofweek']
df.head(5)
````

````python
X = df.drop('dayofweek', axis=1)
y = df['dayofweek']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=21)
````

## 2. Individual classifiers

1. Train SVM, decision tree and random forest again with the best parameters that you got from the 01 exercise with `random_state=21` for all of them.
2. Evaluate `accuracy`, `precision`, and `recall` for them on the validation set.
3. The result of each cell of the section should look like this:

```
accuracy is 0.87778
precision is 0.88162
recall is 0.87778
```

````python
svm_model = SVC(C=10, gamma='auto', kernel='rbf', probability=True, random_state=21)

svm_metrics = ['accuracy', 'precision_weighted', 'recall_weighted']
svm_scores = cross_validate(svm_model, X_train, y_train, scoring=svm_metrics, n_jobs=-1)

print(f'Accuracy is {svm_scores['test_accuracy'].mean():.5f}')
print(f'Precision is {svm_scores['test_precision_weighted'].mean():.5f}')
print(f'Recall is {svm_scores['test_recall_weighted'].mean():.5f}')
````

````python
tree_model = DecisionTreeClassifier(class_weight='balanced', criterion='gini', max_depth=22, random_state=21)

tree_metrics = ['accuracy', 'precision_weighted', 'recall_weighted']
tree_scores = cross_validate(tree_model, X_train, y_train, scoring=tree_metrics, n_jobs=-1)

print(f'Accuracy is {tree_scores['test_accuracy'].mean():.5f}')
print(f'Precision is {tree_scores['test_precision_weighted'].mean():.5f}')
print(f'Recall is {tree_scores['test_recall_weighted'].mean():.5f}')
````

````python
rf_model = RandomForestClassifier(max_depth=28, criterion='gini', class_weight=None, n_estimators=50, random_state=21)

rf_metrics = ['accuracy', 'precision_weighted', 'recall_weighted']
rf_scores = cross_validate(rf_model, X_train, y_train, scoring=rf_metrics, n_jobs=-1)

print(f'Accuracy is {rf_scores['test_accuracy'].mean():.5f}')
print(f'Precision is {rf_scores['test_precision_weighted'].mean():.5f}')
print(f'Recall is {rf_scores['test_recall_weighted'].mean():.5f}')
````

## 3. Voting classifiers

1. Using `VotingClassifier` and the three models that you have just trained, calculate the `accuracy`, `precision`, and `recall` on the validation set.
2. Play with the other parameteres.
3. Calculate the `accuracy`, `precision` and `recall` on the test set for the model with the best weights in terms of accuracy (if there are several of them with equal values, choose the one with the higher precision).

````python
voting_cf = VotingClassifier(estimators=[('svm', svm_model), ('tree', tree_model), ('rf', rf_model)], n_jobs=-1)

metrics = ['accuracy', 'precision_weighted', 'recall_weighted']
scores = cross_validate(voting_cf, X_train, y_train, scoring=metrics, n_jobs=-1)

print(f'Accuracy is {scores['test_accuracy'].mean():.5f}')
print(f'Precision is {scores['test_precision_weighted'].mean():.5f}')
print(f'Recall is {scores['test_recall_weighted'].mean():.5f}')
````

````python
params = {
    'voting': ['soft', 'hard'],
    'weights': [[i, j, k] for i in range(1, 5) for j in range(1, 5) for k in range(1, 5)]
}
````

````python
search_voting = GridSearchCV(VotingClassifier(estimators=[('svm', svm_model), ('tree', tree_model), ('rf', rf_model)], 
                                              n_jobs=-1), param_grid=params)
search_voting.fit(X_train, y_train)
````

````python
search_voting.best_params_
````

````python
voting_cf_best = VotingClassifier(estimators=[('svm', svm_model), ('tree', tree_model), ('rf', rf_model)], n_jobs=-1, voting='hard', weights=[2, 3, 2])

metrics = ['accuracy', 'precision_weighted', 'recall_weighted']
scores_soft = cross_validate(voting_cf_best, X_train, y_train, scoring=metrics, n_jobs=-1)

print(f'Accuracy is {scores_soft['test_accuracy'].mean():.5f}')
print(f'Precision is {scores_soft['test_precision_weighted'].mean():.5f}')
print(f'Recall is {scores_soft['test_recall_weighted'].mean():.5f}')
````

````python
voting_cf_best.fit(X_train, y_train)
voting_predictions = voting_cf_best.predict(X_test)

voting_accuracy = accuracy_score(y_test, voting_predictions)
voting_precision = precision_score(y_test, voting_predictions, average='weighted')
voting_recall = recall_score(y_test, voting_predictions, average='weighted')

print(f'Accuracy is {voting_accuracy:.5f}')
print(f'Precision is {voting_precision:.5f}')
print(f'Recall is {voting_recall:.5f}')
````

## 5. Stacking classifiers

1. To achieve reproducibility in this case you will have to create an object of cross-validation generator: `StratifiedKFold(n_splits=n, shuffle=True, random_state=21)`, where `n` you will try to optimize (the details are below).
2. Using `StackingClassifier` and the three models that you have recently trained, calculate the `accuracy`, `precision` and `recall` on the validation set, try different values of `n_splits` `[2, 3, 4, 5, 6, 7]` in the cross-validation generator and parameter `passthrough` in the classifier itself,
3. Calculate the `accuracy`, `precision`, and `recall` for the model with the best parameters (in terms of accuracy) on the test set (if there are several of them with equal values, choose the one with the higher precision). Use `final_estimator=LogisticRegression(solver='liblinear')`.

````python
estimators = [
    ('svm', svm_model), 
    ('tree', tree_model), 
    ('rf', rf_model)
]
````

````python
best = [0, 0]

for i in range(2, 8):
    stacking_model = StackingClassifier(estimators=estimators, passthrough=True, cv=StratifiedKFold(n_splits=i))
    
    metrics = ['accuracy', 'precision_weighted', 'recall_weighted']
    stacking_scores = cross_validate(stacking_model, X_train, y_train, scoring=metrics, n_jobs=-1)
    
    if stacking_scores['test_accuracy'].mean() > best[0]:
        best[0] = stacking_scores['test_accuracy'].mean()
        best[1] = i
    
    print(f'n_splits={i}')
    print(f'Accuracy is {stacking_scores['test_accuracy'].mean():.5f}')
    print(f'Precision is {stacking_scores['test_precision_weighted'].mean():.5f}')
    print(f'Recall is {stacking_scores['test_recall_weighted'].mean():.5f}\n')

print(f'Best score is {best[0]:.4f} for n_splits = {best[1]}')
````

````python
best_stacking_model = StackingClassifier(estimators=estimators, passthrough=True, cv=StratifiedKFold(n_splits=4), 
                                         final_estimator=OneVsRestClassifier(LogisticRegression(solver='liblinear')))
best_stacking_model.fit(X_train, y_train)

stacking_predictions = best_stacking_model.predict(X_test)
stacking_accuracy = accuracy_score(y_test, stacking_predictions)
stacking_precision = precision_score(y_test, stacking_predictions, average='weighted')
stacking_recall = recall_score(y_test, stacking_predictions, average='weighted')

print(f'Accuracy is {stacking_accuracy:.5f}')
print(f'Precision is {stacking_precision:.5f}')
print(f'Recall is {stacking_recall:.5f}')
````
